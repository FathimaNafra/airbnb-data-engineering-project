```python
import pandas as pd
```

```python
listings = pd.read_csv("../data/raw/london/listings.csv")
calendar = pd.read_csv("../data/raw/london/calendar.csv")
reviews = pd.read_csv("../data/raw/london/reviews.csv")
neighbourhoods = pd.read_csv("../data/raw/london/neighbourhoods.csv")
```

```python
print("Listings:", listings.shape)
print("Calendar:", calendar.shape)
print("Reviews:", reviews.shape)
print("Neighbourhoods:", neighbourhoods.shape)
```

```python
#Dataset Sizes
for name, df in {
    "listings": listings,
    "calendar": calendar,
    "reviews": reviews,
    "neighbourhoods": neighbourhoods
}.items():

    print(name)
    print(df.shape)
```
